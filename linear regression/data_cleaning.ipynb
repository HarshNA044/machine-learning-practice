# Basic Data Cleaning

## Copying Dataframe
We copy a DataFrame—usually using df_cleaned = df.copy()—to protect the original raw data from accidental alterations during the cleaning and preprocessing phases.

- **Creates a Safety Net**: If you make a mistake while cleaning data (like accidentally wiping out a column or filling missing values incorrectly), you can easily revert to the original DataFrame without having to re-download or re-load the raw dataset into memory.

- **Prevents SettingWithCopyWarning**: Pandas often flags a warning when you modify a sliced portion of a DataFrame instead of an explicit copy. Creating a dedicated copy tells Pandas to create a brand new, independent object in your computer's memory, avoiding this common error.

- **Preserves the Audit Trail**: Keeping a copy allows you to compare your "before" and "after" datasets side-by-side during your Exploratory Data Analysis (EDA) to verify that your transformations worked exactly as intended.

### Manual Feature Selection (or Domain-Knowledge-Based Feature Selection)
Our target variable is SalesAmount. To build a mathematically sound linear regression model, we must select our independent variables ($X$) based on a core machine learning rule: The model can only look at information known before the transaction takes place.
#### 1. Columns to KEEP (Model Features)

* **ListPrice**: The baseline sticker price on the shelf. This provides the primary linear anchor for calculating final sales revenue.
* **Category & Subcategory**: Captures product tier definitions (e.g., Bikes vs Mountain Bikes), setting baseline value expectations.
* **Color**: Accounts for consumer aesthetic preferences that might sway demand or premium pricing.
* **PromotionName**: Captures marketing impacts (e.g., Black Friday deals), which heavily drive fluctuations in transaction totals.
* **SalesRegion**: Accounts for geographic purchasing power and regional demand differences.
* **Orderdate**: Extracted into Order_Month and Order_DayOfWeek to capture seasonal trends and weekend spikes.

#### 2. Columns to DROP (Data Leakage / Cheating)

* **TaxAmount & DiscountAmount**: These are calculated directly from or alongside the final SalesAmount. Leaving them in causes data leakage; the model will simply "reverse-engineer" the math instead of learning real customer behavior.
* **UnitPrice & OrderQuantity**: In pure arithmetic, $UnitPrice \times Quantity = SalesAmount$. Providing both forces the model to memorize a basic multiplication formula rather than build predictive intelligence.
* **Freight**: Shipping and backend logistic costs are determined only after items are weighed and boxed post-sale.

#### 3. Columns to DROP (Administrative & Irrelevant Labels)

* **OrderNumber**: A random tracking string database ID with zero numerical relationship to pricing.
* **ProductName**: We drop ProductName because a linear regression model is just a calculator that only understands numbers, not text. However, converting thousands of unique product names into numeric columns would overwhelm the computer, slow down training, and cause the model to overfit (memorize specific items instead of learning general trends). By dropping the individual names and keeping broader columns like Category, we keep the dataset clean, efficient, and ready for mathematical modeling.
* **Duedate & Shipdate**: Backend operational and delivery timestamps that occur after checkout and do not drive consumer buying decisions.

```python
# dropping irrelevant features
df_cleaned = df_cleaned.drop(columns=['TaxAmount', 'DiscountAmount', 'UnitPrice', 'OrderQuantity', 'Freight', 'OrderNumber', 'ProductName', 'Duedate', 'Shipdate'], axis='columns')
```

## Fixing Inconsistent Formats

The Orderdate column has an inconsistent format because some rows use forward slashes ('1/29/2011') while other rows use dashes ('03-01-2011')

```python
df_cleaned['Orderdate'] = pd.to_datetime(df_cleaned['Orderdate'], format='mixed')
# format=mixed: rows in this column contain dates written in different formats, so read and fix each row individually
```

## Checking Impossible Values

The fastest way to spot impossible numbers across our entire dataset is using df.describe(). This prints a summary table showing the minimum, maximum, and average values for every numeric column.

How to spot an impossible value in the summary:
* Look closely at the min row: If it shows a negative number (e.g., -500), that is impossible.
* Look closely at the max row: If someone accidentally typed an extra zero and made a single order quantity 99999, it will stand out immediately compared to the mean.

```python
df_cleaned.describe()
```

* ListPrice can't be Zero: It is the manufacturer's suggested retail price (the standard sticker price on the shelf). A store does not stock items that officially cost ₹0.00 on the shelf.

For example, if a customer applies a 100% off coupon, wins a store giveaway, or exchanges a broken item under a warranty, they walk away paying exactly ₹0.00.
